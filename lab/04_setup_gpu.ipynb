```python
import tensorflow as tf

tf.__version__
```

```python
# 가용 가능한 장치 목록 확인
devices = tf.config.list_physical_devices()
print("총 인식된 장치:", devices)

# GPU(Metal)가 있는지 확인
gpu_devices = tf.config.list_physical_devices('GPU')
if gpu_devices:
    print(f"✅ 성공! Metal GPU를 사용 중입니다: \n{gpu_devices}")
else:
    print("❌ 아직 GPU를 찾지 못했습니다. 커널을 재시작해 보세요.")
```
